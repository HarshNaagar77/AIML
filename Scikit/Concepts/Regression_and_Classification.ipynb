## Linear Regression


Realtion between an independent and a dependent variable  
Y = m X + b 

```python
X = [[ 1] , [2] , [3] , [4] , [5]]
Y = [ [30] , [50] , [70] , [90] , [100]]
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression().fit( X , Y)

hours = float(input("Enter number of hours studied: "))
print(model.predict([[hours]]))
```

## Classification

Predicts a discrete label like spam / not spam , yes / no etc  
Types : 
1. Logistics regression 
2. K-nearest neighbour
3. Decision Tree

### Logistic Regression

Same like  linear regression but give value as binary 

```python
hours_studied = [[ 1] , [2] , [3] , [4] , [5]]
result = [ 0 , 0 , 0 , 1 , 1]
```

```python
from sklearn.linear_model import LogisticRegression
log_model = LogisticRegression().fit( hours_studied , result)

print(log_model.predict([[12]]))
```

### K nearest neighbour

In K nearest neighbour , it predicts value on the basis of neightbour values

```python
from sklearn.neighbors import KNeighborsClassifier
X = [
    [100 , 10],
    [200 , 12],
    [300 , 14],
    [400 , 16]
]
Y = [ 0 , 0 , 1 , 1]
knearest = KNeighborsClassifier(n_neighbors= 3).fit(X , Y)
size = float(input("Enter size: "))
rooms = float(input("Enter weight: "))
print(knearest.predict([[ size , rooms]])[0])
```

Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier
X = [
    [100 , 10],
    [200 , 12],
    [300 , 14],
    [400 , 16]  
]
Y = [ 0 , 0 , 1 , 1]
dtree = DecisionTreeClassifier().fit(X , Y)
size = float(input("Enter size: "))
rooms = float(input("Enter weight: "))

prdict = dtree.predict([[ size , rooms]])[0]
if prdict == 0:
    print("Small House")
else:
    print("Large House")

```
