# **Model Metrics and Evaluation**

It is used to evaluate the performance of model.  
It includes mainly 4 metrics :  
1. Sklearn metrics
2. Regression metrics
3. Confusion metrics
4. Classification metrics

## **Classification metrics**

It includes :
1. **Accuracy**  =>  The ratio of correctly predicted samples to the total samples.  
2. **Precision**  =>   Out Of all the things the model said “yes” to, how many were truly “yes".
3. **Recall**  =>  Out Of all the real “yes” cases, how many the model correctly found.
4. **F1 score**  =>  A single score that shows how well precision and recall balance together.

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score 

x_true = [0, 1, 1, 0, 1, 1, 0, 0, 1, 0]
x_pred = [0, 0, 1, 1, 1, 1, 1, 0, 1, 0]

accuracy = accuracy_score(x_true, x_pred)
precision = precision_score(x_true, x_pred)
recall = recall_score(x_true, x_pred)
f1 = f1_score(x_true, x_pred)
print(f"Accuracy: {accuracy}")
print(f"Precision: {precision}")
print(f"Recall: {recall}")
print(f"F1 Score: {f1}")
```


## **Confusion Metrics**

A table that shows what the model predicted correctly and what it got wrong for each class.

1. **True Positive (TP):** The model predicted “yes,” and it was actually “yes.”  
2. **True Negative (TN):** The model predicted “no,” and it was actually “no.”  
3. **False Positive (FP):** The model predicted “yes,” but it was actually “no.”  
4. **False Negative (FN):** The model predicted “no,” but it was actually “yes.”  

      <img src="https://glassboxmedicine.com/wp-content/uploads/2019/02/confusion-matrix.png?w=816" width="300">


```python
from sklearn.metrics import confusion_matrix
y_true = [0, 1, 1, 0, 1, 1, 0, 0, 1, 0]
y_pred = [0, 0, 1, 1, 1, 1, 1, 0, 1, 0]
cm = confusion_matrix(y_true, y_pred)
print("Confusion Matrix:")
print(cm)
```

## **Regression Metrics**

Regression metrics tell us **how close** our model’s predicted values are to the actual values.

### **Error Metrics - MAE, MSE, RMSE (Simple Explanation With Steps)**

#### **Example Data**
Student marks (Predicted vs Actual):

- A: 90 vs 95 → error = -5  
- B: 50 vs 40 → error = 10  
- C: 66 vs 61 → error = 5  
- D: 80 vs 85 → error = -5  

Errors = -5, 10, 5, -5

---

### **1. Mean Absolute Error (MAE)**  
**Simple meaning:** Average of how far predictions were from actual marks.

#### **Steps**
1. **Find difference** (Predicted - Actual): -5, 10, 5, -5  
2. **Remove negative signs** (take absolute): 5, 10, 5, 5  
3. **Add all values:** 5 + 10 + 5 + 5 = 25  
4. **Divide by total students (4):**  
   MAE = 25 / 4 = **6.25**

---

### **2. Mean Squared Error (MSE)**  
**Simple meaning:** Same as MAE, but big mistakes get punished more.

#### **Steps**
1. **Find difference:** -5, 10, 5, -5  
2. **Square each value:**  
   25, 100, 25, 25  
3. **Add them:** 25 + 100 + 25 + 25 = 175  
4. **Divide by 4:**  
   MSE = 175 / 4 = **43.75**

---

### **3. Root Mean Squared Error (RMSE)**  
**Simple meaning:** Just the **square root** of MSE so the value is back in mark scale.

#### **Steps**
1. **Take MSE:** 43.75  
2. **Take square root:**  
   RMSE = √43.75 ≈ **6.61**


```python
from sklearn.metrics import mean_absolute_error , mean_squared_error
import math
import numpy as np
marks_predicted = [78, 85, 92, 88, 76]
marks_actual = [80, 83, 91, 87, 79]
mae = mean_absolute_error(marks_actual, marks_predicted)
mse = mean_squared_error(marks_actual, marks_predicted)

rmse = math.sqrt(mse)
print(f"Mean Absolute Error: {mae}")
print(f"Mean Squared Error: {mse}")
print(f"Root Mean Squared Error: {rmse}")
```
