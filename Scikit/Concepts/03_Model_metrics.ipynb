# Model Metrics and Evaluation

It is used to evaluate the performance of model.  
It includes mainly 4 metrics :  
1. Sklearn metrics
2. Regression metrics
3. Confusion metrics
4. Classification metrics

## Classification metrics

It includes :
1. **Accuracy**  =>  The ratio of correctly predicted samples to the total samples.  
2. **Precision**  =>   Out Of all the things the model said “yes” to, how many were truly “yes".
3. **Recall**  =>  Out Of all the real “yes” cases, how many the model correctly found.
4. **F1 score**  =>  A single score that shows how well precision and recall balance together.

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score 

x_true = [0, 1, 1, 0, 1, 1, 0, 0, 1, 0]
x_pred = [0, 0, 1, 1, 1, 1, 1, 0, 1, 0]

accuracy = accuracy_score(x_true, x_pred)
precision = precision_score(x_true, x_pred)
recall = recall_score(x_true, x_pred)
f1 = f1_score(x_true, x_pred)
print(f"Accuracy: {accuracy}")
print(f"Precision: {precision}")
print(f"Recall: {recall}")
print(f"F1 Score: {f1}")
```


## Confusion Metrics

A table that shows what the model predicted correctly and what it got wrong for each class.

1. **True Positive (TP):** The model predicted “yes,” and it was actually “yes.”  
2. **True Negative (TN):** The model predicted “no,” and it was actually “no.”  
3. **False Positive (FP):** The model predicted “yes,” but it was actually “no.”  
4. **False Negative (FN):** The model predicted “no,” but it was actually “yes.”
