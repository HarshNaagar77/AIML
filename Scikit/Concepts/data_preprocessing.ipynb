## Encode Categorical values

Categorical data means data which is according to categories like cars , fruits etc. 

Models cannot learn on categorical data so we have to convert them to numerical data.

There are mainly 2 ways to do it:
1. labelEncoder
2. one-hot encoder

### Lable Encoder
