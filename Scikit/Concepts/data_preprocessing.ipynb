# Data preprocessing

There are mainly four phases:
1. Data Cleaning
2. Encode Categroical Values
3. Feature scaling
4. Splitting

# Data Cleaning

```python
import pandas as pd 
```

```python
data = {
     "name" : ["Alice", "Bob", "Charlie" , "Max"],
    "age" : [25, 30 , None , 32 ],
    "city" : ["New York", "Los Angeles", "Chicago" , None]
}
df = pd.DataFrame(data)
print(df)
```

```python
cleanDataAfterDropping = df.dropna()
print("\nData after removing missing values:\n", cleanDataAfterDropping)
```

```python
cleanDataAfterFilling = df.fillna({
    "age": df["age"].mean(),
    "city": "Unknown"
})
print("\nData after filling missing values:\n", cleanDataAfterFilling)
```

# Encode Categorical values

Categorical data means data which is according to categories like cars , fruits etc. 

Models cannot learn on categorical data so we have to convert them to numerical data.

There are mainly 2 ways to do it:
1. labelEncoder
2. one-hot encoder

### Lable Encoder

```python
from sklearn.preprocessing import LabelEncoder
df["Labeled_city"] = LabelEncoder().fit_transform(cleanDataAfterFilling["city"])
print("\nEncoded city data:\n", df)
```

### One-hot Encoder

```python
df_encoded = pd.get_dummies(df ,columns=["city"])
print("\nOne-Hot Encoded city data:\n", df_encoded)
```

# Feature Scaling

Used to scale large range of values to a standard range like 0 to 1 etc

```python
# StandardScaler is used when want values to be around zero .
from sklearn.preprocessing import StandardScaler

scaled_data = StandardScaler().fit_transform(cleanDataAfterFilling[["age"]])
print("\nScaled age data:\n", scaled_data)
```

```python
# MinMaxScaler is used when want values between a specific range (default 0 to 1).
from sklearn.preprocessing import MinMaxScaler
minmax_scaled_data = MinMaxScaler().fit_transform(cleanDataAfterFilling[["age"]])
print("\nMin-Max Scaled age data:\n", minmax_scaled_data)
```

# Dataset Splitting

Splitting dataset for testing and training data

```python
import random
df["Marks"] = [ random.randint(0,100) for _ in range(len(df)) ]
```

```python
from sklearn.model_selection import train_test_split
X = df[["age"]]
Y = df["Marks"]

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
print("\nX_train:\n", X_train)
print("\nX_test:\n", X_test)
print("\nY_train:\n", Y_train)
print("\nY_test:\n", Y_test)
```
